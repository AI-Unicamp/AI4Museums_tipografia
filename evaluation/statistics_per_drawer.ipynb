```python
import pandas as pd



# Se quiser salvar o resultado em um novo CSV:
# estatisticas.to_csv('estatisticas_pastas.csv', index=False)
```

```python
def count_good_crops(path):
# 1. Carregar o arquivo CSV
    df = pd.read_csv(path)

    # 2. Agrupar por gaveta ('folder_path') e recorte ('image_name') para somar os votos 'is_good'
    votos_por_recorte = df.groupby(['folder_path', 'image_name'])['is_good'].sum().reset_index()

    # 3. Criar uma coluna indicando se o recorte foi aprovado (pelo menos 2 votos)
    votos_por_recorte['aprovado'] = (votos_por_recorte['is_good'] >= 2).astype(int)

    # 4. Agrupar apenas por gaveta ('folder_path') para obter as estatísticas finais
    estatisticas_gaveta = votos_por_recorte.groupby('folder_path').agg(
        total_recortes=('image_name', 'count'),
        recortes_bons=('aprovado', 'sum')
    ).reset_index()

    # 5. Calcular a porcentagem de recortes bons
    estatisticas_gaveta['porcentagem_bons (%)'] = ((estatisticas_gaveta['recortes_bons'] / estatisticas_gaveta['total_recortes']) * 100).round(2)

    # 5. Exibir o resultado
    print(estatisticas_gaveta)

    # Salvar o DataFrame em um arquivo CSV, sem incluir os índices das linhas
    estatisticas_gaveta.to_csv(f'stats_{path}', index=False)
```

```python
count_good_crops("annotations_batch_1.csv")
```
